### 4.3 Windowing the EDA Signal

The chest EDA signal is divided into 60-second windows
with a 30-second step.

Each window contains 42,000 samples because the
sampling frequency is 700 Hz.

Each window receives a stress or non-stress label.

```python

import numpy as np

# Convert EDA data into a one-dimensional array
eda = np.asarray(s2["signal"]["chest"]["EDA"]).flatten()

# Get the experimental labels
labels = np.asarray(s2["label"]).flatten()

# Chest sensor sampling frequency
FS = 700

# Window size = 60 seconds
WIN = 60 * FS

# Step size = 30 seconds
STEP = 30 * FS

# Store windows and labels
X_windows = []
y_windows = []

for i in range(0, len(eda) - WIN + 1, STEP):

    window = eda[i:i + WIN]

    # Use the label at the centre of the window
    label = labels[i + WIN // 2]

    X_windows.append(window)
    y_windows.append(label)

X_windows = np.array(X_windows)
y_windows = np.array(y_windows)

print("Total windows:", len(X_windows))
print("Shape:", X_windows.shape)
print("Labels:", np.unique(y_windows))
```

```python

# Keep only baseline (1) and stress (2)
valid = np.isin(y_windows, [1, 2])

X_clean = X_windows[valid]

# Convert baseline to 0 and stress to 1
y_binary = (y_windows[valid] == 2).astype(int)

print("Valid windows:", len(X_clean))
print("Stress windows:", np.sum(y_binary == 1))
print("Non-stress windows:", np.sum(y_binary == 0))
print("Missing values:", np.isnan(X_clean).sum())
```


### 4.4 Preparing Data from Multiple Participants

In this step, participant data files are identified
from the WESAD dataset.

The objective is to prepare data from multiple
participants for subject-wise machine learning
evaluation.

Participants will be separated during model
evaluation to prevent data leakage.


```python

import zipfile
import re
import os

dataset_path = "/content/drive/MyDrive/Speculative_Futures_Project/data/WESAD.zip"

# Open the WESAD ZIP dataset
with zipfile.ZipFile(dataset_path, "r") as zip_ref:

    # Find all participant pickle files
    participant_files = [
        name for name in zip_ref.namelist()
        if re.search(r"(?:^|/)(S\d+)/(S\d+)\.pkl$", name)
        and name.split("/")[-2] ==
            name.split("/")[-1].replace(".pkl", "")
    ]

# Sort files by participant number
participant_files = sorted(
    participant_files,
    key=lambda name: int(
        name.split("/")[-1].replace("S", "").replace(".pkl", "")
    )
)

print("Total participant files:", len(participant_files))

print("\nParticipant files:")
for file in participant_files:
    print(file)
```

```python

# Verify participant files

print("Number of participants:", len(participant_files))

if len(participant_files) == 15:
    print("SUCCESS: All 15 participant files found!")
else:
    print("CHECK: Expected 15 participants.")
    print("Please inspect the ZIP file structure.")
```


### 4.5 Processing All Participants

In this step, EDA signals from all WESAD participants
are processed using 60-second windows and a
30-second step.

Each window is classified as baseline or stress.

Statistical features are extracted and saved for
subsequent machine learning experiments.

All participants will remain identifiable by their
subject IDs to support grouped cross-validation.


```python

import numpy as np
import pandas as pd
import pickle
import zipfile
import io
import os

FS = 700
WIN = 60 * FS
STEP = 30 * FS

rows = []

with zipfile.ZipFile(dataset_path, "r") as z:

    for filename in participant_files:

        subject = os.path.basename(filename).replace(".pkl", "")

        print("Processing:", subject)

        # Load trusted WESAD participant file
        with z.open(filename) as f:
            data = pickle.load(
                io.BytesIO(f.read()),
                encoding="latin1"
            )

        eda = np.asarray(
            data["signal"]["chest"]["EDA"]
        ).flatten()

        labels = np.asarray(data["label"]).flatten()

        # Create 60-second windows
        for start in range(0, len(eda) - WIN + 1, STEP):

            end = start + WIN
            window = eda[start:end]
            window_labels = labels[start:end]

            # Keep only windows entirely within
            # baseline (1) or stress (2)
            if not np.all(window_labels == window_labels[0]):
                continue

            original_label = int(window_labels[0])

            if original_label not in [1, 2]:
                continue

            if not np.isfinite(window).all():
                continue

            binary_label = 1 if original_label == 2 else 0

            diff = np.diff(window)

            rows.append({
                "subject": subject,
                "window_index": start // STEP,
                "eda_mean": float(np.mean(window)),
                "eda_std": float(np.std(window)),
                "eda_min": float(np.min(window)),
                "eda_max": float(np.max(window)),
                "eda_rms": float(
                    np.sqrt(np.mean(diff ** 2))
                ),
                "label": binary_label
            })

        # Free participant data before loading next one
        del data, eda, labels

F = pd.DataFrame(rows)

print("\nProcessing complete!")
print("Total windows:", len(F))
print("Total subjects:", F["subject"].nunique())

display(F.head())
```

```python

print("Dataset shape:", F.shape)

print("\nParticipant count:")
print(F["subject"].nunique())

print("\nClass distribution:")
print(F["label"].value_counts())

print("\nMissing values:")
print(F.isna().sum())

print("\nDuplicate subject-window pairs:")
print(F.duplicated(
    subset=["subject", "window_index"]
).sum())
```


### 4.6 Subject-Wise Splitting and Leakage Verification

To prevent data leakage, the dataset is divided
using GroupKFold cross-validation.

Each participant is assigned to either the
training group or the testing group within a fold.

No participant is allowed to appear in both
groups within the same fold.

This ensures that the machine learning model
is evaluated using unseen participants.


```python

import pandas as pd

file_path = (
    "/content/drive/MyDrive/"
    "Speculative_Futures_Project/data/clean_features.csv"
)

F = pd.read_csv(file_path)

print("Dataset loaded successfully!")

print("Total rows:", len(F))
print("Total participants:", F["subject"].nunique())

display(F.head())
```

```python

from sklearn.model_selection import GroupKFold
import numpy as np

# Select the observable feature columns
feature_cols = [
    "eda_mean",
    "eda_std",
    "eda_min",
    "eda_max",
    "eda_rms"
]

X = F[feature_cols]
y = F["label"]
groups = F["subject"]

# Create five subject-wise folds
gkf = GroupKFold(n_splits=5)

fold_splits = list(gkf.split(X, y, groups))

for fold, (train_idx, test_idx) in enumerate(
    fold_splits, start=1
):
    train_subjects = set(groups.iloc[train_idx])
    test_subjects = set(groups.iloc[test_idx])

    overlap = train_subjects.intersection(test_subjects)

    print(f"\nFold {fold}")
    print("Training subjects:", sorted(train_subjects))
    print("Testing subjects:", sorted(test_subjects))
    print("Subject overlap:", overlap)

    assert len(overlap) == 0

print("\nSUCCESS: No subject leakage detected!")
```

```python

print("Class distribution:")
print(F["label"].value_counts())

print("\nClass percentages:")
print(F["label"].value_counts(normalize=True) * 100)

print("\nMissing values:")
print(F.isna().sum())

print("\nDuplicate subject-window pairs:")
print(
    F.duplicated(
        subset=["subject", "window_index"]
    ).sum()
)

print("\nClass labels in each fold:")

for fold, (train_idx, test_idx) in enumerate(
    fold_splits, start=1
):
    train_labels = set(y.iloc[train_idx])
    test_labels = set(y.iloc[test_idx])

    print(
        f"Fold {fold}: "
        f"Train={sorted(train_labels)}, "
        f"Test={sorted(test_labels)}"
    )
```

## Step 5: Observable Feature Engineering

### 5.1 Introduction

Feature engineering transforms raw physiological

1. EDA Mean
2. EDA Standard Deviation

```python

import pandas as pd
import numpy as np

file_path = (
    "/content/drive/MyDrive/"
    "Speculative_Futures_Project/data/clean_features.csv"
)

F = pd.read_csv(file_path)

feature_cols = [
    "eda_mean",
    "eda_std",
    "eda_min",
    "eda_max",
    "eda_rms"
]

print("Feature dataset loaded successfully!")

print("Dataset shape:", F.shape)

display(F.head())
```

```python

feature_dictionary = pd.DataFrame({
    "Feature": [
        "eda_mean",
        "eda_std",
        "eda_min",
        "eda_max",
        "eda_rms"
    ],
    "Definition": [
        "Mean EDA value within a 60-second window",
        "Standard deviation of EDA values",
        "Minimum EDA value",
        "Maximum EDA value",
        "RMS of consecutive EDA differences"
    ],
    "Purpose": [
        "Overall skin conductance level",
        "Variation in skin conductance",
        "Lowest observed conductance",
        "Highest observed conductance",
        "Short-term EDA signal changes"
    ]
})

display(feature_dictionary)
```

```python

import matplotlib.pyplot as plt

# Calculate feature correlations using raw values
correlation_matrix = F[feature_cols].corr()

print("Feature Correlation Matrix:")
display(correlation_matrix.round(2))

plt.figure(figsize=(8, 6))
plt.imshow(
    correlation_matrix,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(feature_cols)),
    feature_cols,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(feature_cols)),
    feature_cols
)

plt.title("EDA Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python

threshold = 0.8

high_correlations = []

for i in range(len(feature_cols)):
    for j in range(i + 1, len(feature_cols)):

        feature1 = feature_cols[i]
        feature2 = feature_cols[j]

        correlation = correlation_matrix.loc[
            feature1, feature2
        ]

        if abs(correlation) > threshold:
            high_correlations.append({
                "Feature 1": feature1,
                "Feature 2": feature2,
                "Correlation": round(correlation, 3)
            })

high_corr_df = pd.DataFrame(
    high_correlations,
    columns=["Feature 1", "Feature 2", "Correlation"]
)

print("Feature pairs with |correlation| > 0.8:")
display(high_corr_df)
```


## Step 6: Speculative Feature Extraction

### 6.1 Introduction

This step introduces a speculative feature called
surveillance_pressure_2030.

The feature represents hypothetical workplace
surveillance pressure experienced by employees
in the year 2030.

The WESAD dataset does not contain actual
workplace surveillance information.

Therefore, this feature is generated using a
statistical simulation rather than real measurements.

A clipped normal distribution is used with:

- Mean (mu): 0.55
- Standard deviation (sigma): 0.15
- Minimum value: 0
- Maximum value: 1
- Random seed: 42

The purpose is to examine how adding this
hypothetical contextual feature influences
machine learning predictions.

## Step 7: Training and Evaluation

### 7.1 Introduction

This step evaluates the effectiveness of machine learning
models for stress classification using the WESAD dataset.

Two Random Forest models are compared:

1. Baseline Model:
   Uses five observable EDA features.

2. Speculative Model:
   Uses the same five EDA features plus the simulated
   surveillance_pressure_2030 feature.

Both models use identical subject-wise cross-validation
splits and model settings to ensure a fair comparison.

Performance is evaluated using accuracy and F1-score.
Model complexity and error differences across simulated
pressure levels will also be examined.


```python

import pandas as pd
import numpy as np

file_path = (
    "/content/drive/MyDrive/"
    "Speculative_Futures_Project/data/"
    "augmented_feature_matrix.csv"
)

F = pd.read_csv(file_path)

base_features = [
    "eda_mean",
    "eda_std",
    "eda_min",
    "eda_max",
    "eda_rms"
]

spec_features = base_features + [
    "surveillance_pressure_2030"
]

print("Dataset loaded successfully!")
print("Total windows:", len(F))
print("Total participants:", F["subject"].nunique())
print("Baseline features:", len(base_features))
print("Speculative features:", len(spec_features))

display(F.head())
```

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GroupKFold, cross_val_predict
from sklearn.metrics import accuracy_score, f1_score

# Stress labels and participant IDs
y = F["label"]
groups = F["subject"]

# Ensure our dataset is suitable for evaluation
assert F["subject"].nunique() >= 5
assert set(y.unique()) == {0, 1}
assert not F[spec_features].isna().any().any()

# Create identical subject-wise splits for both models
gkf = GroupKFold(n_splits=5)
splits = list(gkf.split(F[base_features], y, groups))

# Check that train and test subjects never overlap
for train_idx, test_idx in splits:
    train_subjects = set(groups.iloc[train_idx])
    test_subjects = set(groups.iloc[test_idx])
    assert train_subjects.isdisjoint(test_subjects)

# Create a function for model evaluation
def evaluate(feature_columns, seed=42):

    model = RandomForestClassifier(
        n_estimators=200,
        random_state=seed,
        n_jobs=-1
    )

    predictions = cross_val_predict(
        model,
        F[feature_columns],
        y,
        cv=splits,
        n_jobs=1
    )

    accuracy = accuracy_score(y, predictions)
    f1 = f1_score(y, predictions, zero_division=0)

    return accuracy, f1, predictions

# Model 1: Baseline
acc_base, f1_base, pred_base = evaluate(
    base_features
)

# Model 2: Speculative
acc_spec, f1_spec, pred_spec = evaluate(
    spec_features
)

print("BASELINE MODEL")
print("Accuracy:", round(acc_base, 4))
print("F1-score:", round(f1_base, 4))

print("\nSPECULATIVE MODEL")
print("Accuracy:", round(acc_spec, 4))
print("F1-score:", round(f1_spec, 4))

print("\nCHANGE IN F1-SCORE")
print(round(f1_spec - f1_base, 4))
```

```python

comparison = pd.DataFrame({
    "Model": [
        "Baseline Random Forest",
        "Speculative Random Forest"
    ],
    "Number of Features": [5, 6],
    "Accuracy": [acc_base, acc_spec],
    "F1-score": [f1_base, f1_spec],
    "Number of Trees": [200, 200]
})

display(comparison.round(4))
```


### 7.9 Fairness Analysis

This analysis investigates whether prediction error rates
differ across low, medium, and high simulated workplace
surveillance pressure levels.

Participants are divided into three groups according to
their simulated surveillance pressure scores.

The error rate is calculated for each group using
cross-validated predictions from the speculative model.

The difference between the highest and lowest group
error rates is reported as the error gap.

These results describe simulated pressure groups, not
fairness across actual demographic characteristics.


```python

import pandas as pd
import numpy as np

# Use predictions already generated in Step 7.4
F["prediction"] = pred_spec

# A value of 1 means the prediction was incorrect
F["error"] = (F["prediction"] != F["label"]).astype(int)

# One pressure value for each participant
subject_pressure = (
    F[["subject", "surveillance_pressure_2030"]]
    .drop_duplicates("subject")
    .sort_values("surveillance_pressure_2030")
    .copy()
)

# Divide participants into three equal-sized groups
subject_pressure["pressure_group"] = pd.qcut(
    subject_pressure["surveillance_pressure_2030"].rank(
        method="first"
    ),
    q=3,
    labels=["Low", "Medium", "High"]
)

# Assign the groups to all windows
group_map = subject_pressure.set_index(
    "subject"
)["pressure_group"]

F["pressure_group"] = F["subject"].map(group_map)

# Calculate the error rate in each group
fairness_results = (
    F.groupby("pressure_group", observed=True)["error"]
    .agg(["mean", "count"])
    .rename(columns={
        "mean": "error_rate",
        "count": "number_of_windows"
    })
)

display(fairness_results)

# Difference between highest and lowest error rate
error_gap = (
    fairness_results["error_rate"].max()
    - fairness_results["error_rate"].min()
)

print("Error gap:", round(error_gap, 4))
print("Error gap percentage points:",
      round(error_gap * 100, 2))
```


### 7.13 Model Complexity and Maintainability

This section compares the computational complexity
and maintainability of the baseline and speculative
Random Forest models.

The evaluation considers:

1. Number of input features
2. Number of decision trees
3. Average decision tree depth
4. Model training time
5. Ease of retraining and reproducibility

Both models use identical Random Forest settings
to ensure a consistent comparison.


```python

import time
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier

complexity_results = []

for model_name, features in [
    ("Baseline", base_features),
    ("Speculative", spec_features)
]:

    # Use only the first training fold
    train_idx, test_idx = splits[0]

    X_train = F.iloc[train_idx][features]
    y_train = F.iloc[train_idx]["label"]

    model = RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    )

    start_time = time.perf_counter()

    model.fit(X_train, y_train)

    training_time = time.perf_counter() - start_time

    tree_depths = [
        tree.tree_.max_depth
        for tree in model.estimators_
    ]

    complexity_results.append({
        "Model": model_name,
        "Features": len(features),
        "Trees": len(model.estimators_),
        "Average Tree Depth": np.mean(tree_depths),
        "Training Time (seconds)": training_time
    })

complexity_df = pd.DataFrame(complexity_results)

print("Model Complexity Comparison:")
display(complexity_df.round(3))
```


## Step 8: Verification and Validation of Results

### 8.1 Introduction

This step evaluates the reliability and reproducibility
of the machine learning results.

The baseline and speculative Random Forest models
are evaluated using three random seeds:

- 42
- 7
- 2026

The same dataset, features, model parameters,
and subject-wise cross-validation splits are used.

Accuracy and F1-score are recorded for each seed.

Mean and standard deviation are calculated to
measure performance and variability across runs.


```python

import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import accuracy_score, f1_score

# We use the same dataset and splits from Step 7
y = F["label"]
groups = F["subject"]

# Re-create the same five-fold subject-wise splits
from sklearn.model_selection import GroupKFold

gkf = GroupKFold(n_splits=5)

splits = list(
    gkf.split(F[base_features], y, groups)
)

# Function to evaluate one model
def evaluate_with_seed(feature_columns, seed):

    model = RandomForestClassifier(
        n_estimators=200,
        random_state=seed,
        n_jobs=-1
    )

    predictions = cross_val_predict(
        model,
        F[feature_columns],
        y,
        cv=splits
    )

    accuracy = accuracy_score(y, predictions)

    f1 = f1_score(
        y,
        predictions,
        zero_division=0
    )

    return accuracy, f1

# Three random seeds
seeds = [42, 7, 2026]

reliability_results = []

for seed in seeds:

    print("Running seed:", seed)

    # Baseline model
    acc_base_seed, f1_base_seed = evaluate_with_seed(
        base_features, seed
    )

    # Speculative model
    acc_spec_seed, f1_spec_seed = evaluate_with_seed(
        spec_features, seed
    )

    reliability_results.append({
        "Seed": seed,
        "Baseline Accuracy": acc_base_seed,
        "Baseline F1": f1_base_seed,
        "Speculative Accuracy": acc_spec_seed,
        "Speculative F1": f1_spec_seed,
        "F1 Difference": f1_spec_seed - f1_base_seed
    })

reliability_df = pd.DataFrame(reliability_results)

print("\nReliability Results:")
display(reliability_df.round(4))
```

```python

# Calculate reliability statistics
reliability_summary = reliability_df.drop(
    columns=["Seed"]
).agg(["mean", "std"])

print("Mean and Standard Deviation:")

display(reliability_summary.round(4))

print("\nBaseline F1:")
print(
    f"{reliability_df['Baseline F1'].mean():.4f} ± "
    f"{reliability_df['Baseline F1'].std():.4f}"
)

print("\nSpeculative F1:")
print(
    f"{reliability_df['Speculative F1'].mean():.4f} ± "
    f"{reliability_df['Speculative F1'].std():.4f}"
)
```

```python

import os
import pandas as pd

project_folder = (
    "/content/drive/MyDrive/"
    "Speculative_Futures_Project"
)

required_files = [
    "data/WESAD.zip",
    "data/clean_features.csv",
    "data/feature_matrix.csv",
    "data/feature_dictionary.csv",
    "data/augmented_feature_matrix.csv",
    "data/hyperparameter_justification.csv",
    "data/sensitivity_summary.csv",
    "results/baseline_vs_speculative.csv",
    "results/fairness_analysis.csv",
    "results/model_sensitivity_analysis.csv",
    "results/model_complexity.csv",
    "results/final_model_comparison.csv",
    "results/reliability_runs.csv",
    "results/reliability_summary.csv",
    "results/validity_checklist.csv",
    "results/literature_benchmark.csv",
    "results/participant_performance.csv"
]

file_results = []

for filename in required_files:
    path = os.path.join(project_folder, filename)

    file_results.append({
        "File": filename,
        "Status": "FOUND" if os.path.isfile(path)
                  else "MISSING"
    })

verification_df = pd.DataFrame(file_results)

display(verification_df)

print(
    "Files found:",
    (verification_df["Status"] == "FOUND").sum()
)

print(
    "Files missing:",
    (verification_df["Status"] == "MISSING").sum()
)
```
